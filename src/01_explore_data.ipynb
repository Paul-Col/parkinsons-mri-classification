```python
# Set the project root to the parent of the current working directory
PROJECT_ROOT = Path.cwd().parent

# Create the path to the raw dataset folder
DATA_DIR = PROJECT_ROOT / "data" / "raw"

# Create a list containing all files inside the raw dataset folder
files = [
    path for path in DATA_DIR.rglob("*")
    if path.is_file()
]

# Display the total number of files found
print(f"Total number of files: {len(files)}")
```

```python
# Create a counter to store how many files exist for each file extension
extensions = Counter()

# Loop through every file found in the dataset
for file in files:

    # Convert the file extension to lowercase and add it to the counter
    extensions[file.suffix.lower()] += 1

# Display a heading for the file type summary
print("\nFile types:")
print("-" * 30)

# Loop through each file extension and display how many files were found
for extension, count in extensions.items():
    print(f"{extension}: {count}")
```

```python
# Define a function that returns the correct file extension
def get_extension(path):
    # Convert the filename to lowercase so extension checks are consistent
    name = path.name.lower()

    # Check specifically for compressed NIfTI files
    # Path.suffix would only return ".gz", so this handles ".nii.gz" correctly
    if name.endswith(".nii.gz"):
        return ".nii.gz"

    # Return the normal file extension for all other files
    return path.suffix.lower()


# Count how many files exist for each file type
extensions = Counter(
    get_extension(file)
    for file in files
)

# Display a heading for the file type summary
print("\nFILE TYPES")
print("=" * 60)

# Sort the file extensions alphabetically and display each count
for extension, count in sorted(extensions.items()):
    print(f"{extension:<10} {count}")
```

```python
# Display a heading for the top-level dataset folders
print("\nTOP LEVEL DIRECTORIES")
print("=" * 60)

# Loop through each item directly inside the raw dataset directory
for folder in DATA_DIR.iterdir():

    # Only process items that are directories
    if folder.is_dir():

        # Count all files contained in the folder and its subfolders
        file_count = sum(
            1
            for file in folder.rglob("*")
            if file.is_file()
        )

        # Display the folder name and the number of files it contains
        print(f"{folder.name:<30} {file_count}")
```

```python
# Project paths

# Set the project root to the parent of the current working directory
PROJECT_ROOT = Path.cwd().parent

# Create the path to the raw dataset folder
DATA_DIR = PROJECT_ROOT / "data" / "raw"


# Helper functions

# Return the correct file extension for each dataset file
def get_extension(path):

    # Convert the filename to lowercase
    name = path.name.lower()

    # Handle compressed NIfTI MRI files correctly
    if name.endswith(".nii.gz"):
        return ".nii.gz"

    # Return the standard extension for other file types
    return path.suffix.lower()


# Check dataset

# Display the main heading
print("=" * 60)
print("PARKINSON'S DATASET EXPLORATION")
print("=" * 60)

# Display the location of the dataset
print(f"\nDataset directory:\n{DATA_DIR}")

# Check that the raw dataset directory exists
if not DATA_DIR.exists():
    print("\nERROR: Dataset directory does not exist.")
    raise SystemExit


# Find files

# Create a list containing every file in the raw dataset
# including files inside subdirectories
files = [
    path
    for path in DATA_DIR.rglob("*")
    if path.is_file()
]

# Display the total number of files found
print(f"\nTotal files: {len(files)}")


# File types

# Count how many files exist for each file extension
extensions = Counter(
    get_extension(file)
    for file in files
)

# Display the file type summary
print("\nFILE TYPES")
print("-" * 60)

# Sort the extensions alphabetically and display their counts
for extension, count in sorted(extensions.items()):
    print(f"{extension:<15} {count}")


# Top level directories

# Display the main folders contained directly inside data/raw
print("\nTOP LEVEL DIRECTORIES")
print("-" * 60)

# Loop through each item in the raw dataset directory
for folder in DATA_DIR.iterdir():

    # Only process directories
    if folder.is_dir():

        # Count all files inside the folder and its subfolders
        count = sum(
            1
            for path in folder.rglob("*")
            if path.is_file()
        )

        # Display the folder name and number of files
        print(f"{folder.name:<30} {count}")


# Example files

# Display the first 20 files found in the dataset
print("\nFIRST 20 FILES")
print("-" * 60)

# Print each path relative to the raw dataset folder
for file in files[:20]:
    print(file.relative_to(DATA_DIR))
```
